```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("ttft_summary.csv")

plt.figure(figsize=(6,5))
plt.bar(df["model"], df["average"])
plt.title("Average TTFT by Model")
plt.xlabel("Model")
plt.ylabel("Average TTFT (ms)")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.savefig("ttft_summary.png", dpi=300, bbox_inches="tight")
plt.tight_layout()
plt.show()
```

```python


df = pd.read_csv("ttft_results.csv")

for model in df["model"].unique():
    temp = df[df["model"] == model]
    plt.figure(figsize=(6,5))
    plt.plot(temp["run"], temp["ttft_ms"], marker="o")
    plt.title(f"TTFT Across Runs - {model}")
    plt.xlabel("Run")
    plt.ylabel("TTFT (ms)")
    plt.xticks(temp["run"])
    plt.grid(True)
    plt.tight_layout()
    plt.show()

plt.savefig("ttft_results.png", dpi=300, bbox_inches="tight")
```

```python
df = pd.read_csv("perplexity_results.csv")

plt.figure(figsize=(6,5))
plt.barh(df["Model"], df["Average"])
plt.title("Average Perplexity")
plt.xlabel("Perplexity")

plt.savefig("perplexity_results.png", dpi=300, bbox_inches="tight")
plt.tight_layout()
plt.show()
```

```python


df = pd.read_csv("bleu_results.csv")
bleu_avg = df.groupby("model")["corpus_bleu"].mean()

plt.figure(figsize=(6,5))
plt.bar(bleu_avg.index, bleu_avg.values, width=0.5)

plt.title("Average BLEU Score by Model")
plt.xlabel("Model")
plt.ylabel("Average BLEU Score")
plt.grid(axis="y", linestyle="--", alpha=0.5)

for i, value in enumerate(bleu_avg.values):
    plt.text(i, value + 0.05, f"{value:.2f}", ha="center", fontsize=10)

plt.savefig("bleu_bargraph.png", dpi=300, bbox_inches="tight")
plt.tight_layout()
plt.show()
```
