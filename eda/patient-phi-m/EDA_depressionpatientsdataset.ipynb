```python
import json
import random
from copy import deepcopy

# Load JSON
with open("Patient Psi CM Dataset.json", "r", encoding="utf-8") as f:
    data = json.load(f)

filled_data = deepcopy(data)


patient_pool = {}
global_pool = {}

for record in data:
    patient = record.get("name")

    if patient not in patient_pool:
        patient_pool[patient] = {}

    for key, value in record.items():

        valid = False

        if value is None:
            valid = False
        elif isinstance(value, str):
            valid = value.strip() != ""
        elif isinstance(value, list):
            valid = len(value) > 0
        else:
            valid = True

        if valid:
            patient_pool[patient].setdefault(key, []).append(value)
            global_pool.setdefault(key, []).append(value)
```

```python
patient_pool
```

```python

# Fill missing values
for record in filled_data:

    patient = record["name"]

    for key, value in record.items():

        missing = (
            value is None or
            (isinstance(value, str) and value.strip() == "") or
            (isinstance(value, list) and len(value) == 0)
        )

        if missing:

            # First preference: same patient
            if key in patient_pool[patient]:
                record[key] = deepcopy(random.choice(patient_pool[patient][key]))

            # Second preference: any patient
            elif key in global_pool:
                record[key] = deepcopy(random.choice(global_pool[key]))

```

```python

# Save new JSON
with open("Patient_Psi_CM_Dataset_Filled.json", "w", encoding="utf-8") as f:
    json.dump(filled_data, f, indent=4, ensure_ascii=False)

print("Dataset saved as Patient_Psi_CM_Dataset_Filled.json")
```

```python
import json
import pandas as pd

with open("Patient Psi CM Dataset.json","r",encoding="utf-8") as f:
    data = json.load(f)

scores = []

for patient in data:

    score = 0

    helpless = patient.get("helpless_belief",
                 patient.get("helpless_belief_current", []))

    unlovable = patient.get("unlovable_belief",
                   patient.get("unlovable_belief_current", []))

    worthless = patient.get("worthless_belief",
                   patient.get("worthless_belief_current", []))

    emotion = " ".join(patient.get("emotion", [])).lower()
    behavior = patient.get("behavior","").lower()
    thought = patient.get("auto_thought","").lower()

    if len(helpless) > 0:
        score += 2

    if len(unlovable) > 0:
        score += 2

    if len(worthless) > 0:
        score += 3

    if "sad" in emotion:
        score += 2

    if "lonely" in emotion:
        score += 2

    if "guilty" in emotion:
        score += 2

    if "ashamed" in emotion:
        score += 2

    if "avoid" in behavior:
        score += 2

    if "withdraw" in behavior:
        score += 2

    if "sleep" in behavior:
        score += 1

    if "hopeless" in thought:
        score += 2

    if "never" in thought:
        score += 1

    if "worthless" in thought:
        score += 3

    if "failure" in thought:
        score += 2

    if "can't" in thought or "cannot" in thought:
        score += 1

    scores.append({
        "Name": patient["name"],
        "ID": patient["id"],
        "Score": score
    })

```

```python
df = pd.DataFrame(scores)

df = df.sort_values("Score", ascending=False)

print(df)
```

```python
with open("Depression_Patients.json", "w", encoding="utf-8") as f:
    json.dump(top16_patients, f, indent=4, ensure_ascii=False)

print("Top 16 patients saved successfully.")
print("Total patients saved:", len(top16_patients))
```
